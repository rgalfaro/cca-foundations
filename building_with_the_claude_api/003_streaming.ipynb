```python
# Import the load_dotenv function from the dotenv module
from dotenv import load_dotenv
# Call the load_dotenv function to load environment variables from a .env file
load_dotenv()
```

```python
# Create a client to interact with the API
from anthropic import Anthropic
client = Anthropic()
#Define the model to use for the API calls. You can change this to any model you want to use.
model = "claude-haiku-5-5"  
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

#For this example on streaming Claude's response, we will ignore this function and create a streaming response for the chat function. The streaming response is created by calling the client.messages.create() method with the same parameters as the chat function, but with the stream parameter set to True. The streaming response is stored in the stream variable on the next piece of code in this notebook. The stream variable is then used to iterate over the response and print the text of the assistant's response as it is generated
def chat(messages, system=None):
    #Set parameters for the API call
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        # Commenting out the temperatur as not needed for this example, but can be used to control the randomness of the response. A higher temperature will result in more random responses, while a lower temperature will result in more deterministic responses.
        # If using temperature, remember to add it to the function definition and pass it as an argument when calling the function.
        #"temperature": temperature,
    }
    # If a system message is provided, add it to the parameters
    if system is not None:
        params["system"] = system
    # Make the API call to create a message and return the assistant's response. 
    # client.messages.create() is a method that sends the messages to the API and returns the assistant's response and uses parameters defined in the params dictionary. The response is stored in the message variable.
    message = client.messages.create(**params)
    #return the text of the assistant's response from the message object. The response is stored in the messages.content[0].text attribute. The [0] index is used to access the first message in the messages list, which is the assistant's response.
    return messages.content[0].text
```

```python
# Create an empty list to store the messages in the conversation. The messages list is used to store the user and assistant messages in the conversation. The messages list is passed to the chat function to send the messages to the API and get the assistant's response.
messages = []

#Create a system message to set the context for the conversation. The system message is a dictionary with a "role" key set to "system" and a "content" key set to the text of the system message. The system message is added to the messages list using the add_assistant_message function.
add_user_message(messages, "Write a one sentence description of a fake database")

# Call the chat function to send the messages to the API and get the assistant's response. The response is stored in the answer variable.
# Create a streaming response for the chat function. The streaming response is created by calling the client.messages.create() method with the same parameters as the chat function, but with the stream parameter set to True. The streaming response is stored in the stream variable.
stream = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=messages,
    stream=True,
)
# For loop to iterate over the response and print the text of the assistant's response as it is generated. The stream variable is an iterator that yields events as they are generated by the API. The for loop iterates over the events and prints the text of the assistant's response as it is generated. The event.type attribute is used to check if the event is a response.output_text.delta event, which indicates that the event contains a delta of the assistant's response. The event.delta attribute is used to get the text of the assistant's response from the event.
for event in stream:
    print(event)
```
